```python
from pathlib import Path

ROOT = Path('../')
label_dir = ROOT / 'data/processed/labels/val'
img_dir = ROOT / 'data/processed/images/val'

positive_labels = [f for f in label_dir.glob('*.txt') if f.stat().st_size > 0]

print(f'Found {len(positive_labels)} positive labels')

for f in positive_labels[:5]:
    img_path = img_dir / f.name.replace('.txt', '.png')
    print(img_path.resolve())
```
